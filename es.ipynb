```python
da_stud = ["Iniemen", "Chucks", "Becky", "Alice", "chigozie", "Chris", "Isreal", "Id", "Tony", "Jimmy", "Mary"]
da_presentees = ["Iniemen", "Chucks","chigozie", "Becky", "Alice", "Tony", "Jimmy", "Mary"]
```
